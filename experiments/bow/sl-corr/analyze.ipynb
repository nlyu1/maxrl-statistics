## BoW SL analysis script

Parses study artifacts from `sweep.py`.

```python
import polars as pl
from sweep import CORR_LIST, STUDY_BASE, get_study_folder

from src.experiments.bag_of_words.analysis import (
    BagOfWordsAnalysisConfig,
    corr_expr,
    mse_expr,
    rsq_expr,
)
```

```python
studies = {
    f"corr_{corr:.3f}": get_study_folder(study_base=STUDY_BASE, corr=corr)
    for corr in CORR_LIST
}
analysis = BagOfWordsAnalysisConfig.from_studies(studies)
print(f"{len(analysis.studies)}/{len(CORR_LIST)} experiments started")
analysis.get_metric_dataframe()
```

### Best-epoch summaries

One point per study at the epoch that maximizes (for corr) or minimizes (for mse) the y-axis metric.

```python
epoch_axis = pl.col("epoch").alias("epoch")

analysis.xy_plots([
    (
        epoch_axis,
        rsq_expr(split="train", y="target"),
        None,
    ),
    (
        epoch_axis,
        rsq_expr(split="train", y="ground_truth"),
        None,
    ),
    (
        corr_expr(split="train", y="target"),
        corr_expr(split="train", y="ground_truth"),
        None,
    ),
])
```

```python
analysis.xy_plots([
    (
        rsq_expr(split="train", y="target"),
        corr_expr(split="train", y="ground_truth"),
        True,
    ),
    (
        corr_expr(split="train", y="target"),
        corr_expr(split="val", y="ground_truth"),
        True,
    ),
])
```
